## SQL Analysis

### 1. Find Books in the Classic Fiction Genre

**Business question:**  
Which books belong to the Classic Fiction genre?

```python
query = """
SELECT *
FROM books
WHERE genre = 'Classic Fiction';
"""

execute_query(query)
```

### 2. Find Books Above the Average List Price

**Business question:**  
Which books have a list price higher than the average list price of all books?

```python
query = """
SELECT *
FROM books
WHERE list_price > (
    SELECT AVG(list_price)
    FROM books
);
"""

execute_query(query)
```

### 3. Total List Price by Location

**Business question:**  
What is the total list price of books at each current location?

```python
query = """
SELECT
    current_location,
    SUM(list_price) AS total_value
FROM books
GROUP BY current_location
ORDER BY total_value DESC;
"""

execute_query(query)
```

### 4. Find High-Value Transactions

**Business question:**  
Which transactions have a sale price higher than the average transaction sale price?

```python
query = """
SELECT *
FROM transactions
WHERE sale_price > (
    SELECT AVG(sale_price)
    FROM transactions
)
ORDER BY sale_price DESC;
"""

execute_query(query)
```

### 5. Analyze Customer Purchases

**Business question:**  
Which customers have made more than 5 purchases, and how much have they spent in total?

```python
query = """
SELECT
    c.customer_id,
    COUNT(t.transaction_id) AS total_purchases,
    SUM(t.sale_price) AS total_amount_spent
FROM customers AS c
JOIN transactions AS t
    ON c.customer_id = t.customer_id
GROUP BY c.customer_id
HAVING COUNT(t.transaction_id) > 5
ORDER BY total_amount_spent DESC;
"""

execute_query(query)
```

### 6. Combine Customer, Transaction, and Book Data

**Business question:**  
What books did customers purchase, when did they purchase them, where did the purchase occur, and how much did they pay?

```python
query = """
SELECT
    c.customer_id,
    c.preferred_store,
    t.transaction_id,
    t.date_time,
    t.store_location,
    b.book_id,
    b.title,
    b.genre,
    t.sale_price,
    t.payment_method,
    t.is_online
FROM customers AS c
JOIN transactions AS t
    ON c.customer_id = t.customer_id
JOIN books AS b
    ON t.book_id = b.book_id
ORDER BY t.date_time;
"""

execute_query(query)
```

### 7. Top-Selling Book in Each Genre at Each Store

**Business question:**  
What is the top-selling book in each genre at each BookCycle store?

```python
query = """
WITH book_sales AS (
    SELECT
        t.store_location,
        b.genre,
        b.book_id,
        b.title,
        COUNT(t.transaction_id) AS total_sold
    FROM transactions AS t
    JOIN books AS b
        ON t.book_id = b.book_id
    GROUP BY
        t.store_location,
        b.genre,
        b.book_id,
        b.title
),
ranked_books AS (
    SELECT
        store_location,
        genre,
        book_id,
        title,
        total_sold,
        ROW_NUMBER() OVER (
            PARTITION BY store_location, genre
            ORDER BY total_sold DESC
        ) AS rank
    FROM book_sales
)
SELECT
    store_location,
    genre,
    book_id,
    title,
    total_sold
FROM ranked_books
WHERE rank = 1
ORDER BY
    store_location,
    genre;
"""

execute_query(query)
```

### 8. Customers with More Than 5 Purchases

**Business question:**  
Which customers have made more than 5 purchases, and how much have they spent in total?

```python
query = """
SELECT
    c.customer_id,
    COUNT(t.transaction_id) AS total_purchases,
    SUM(t.sale_price) AS total_amount_spent
FROM customers AS c
JOIN transactions AS t
    ON c.customer_id = t.customer_id
GROUP BY c.customer_id
HAVING COUNT(t.transaction_id) > 5
ORDER BY total_amount_spent DESC;
"""

execute_query(query)
```

### Connect to the SQLite Database



```python
import sqlite3
import pandas as pd

# Connect to the BookCycle database
conn = sqlite3.connect('bookcycle.db')
cursor = conn.cursor()

# Function to execute SQL queries and display results
def execute_query(query):
    df = pd.read_sql_query(query, conn)
    display(df)
```

### Create the Books Table

```python
# Create the books table
create_table_query = """
CREATE TABLE IF NOT EXISTS books (
    book_id TEXT PRIMARY KEY,  
    title TEXT,
    author TEXT,
    isbn TEXT,
    genre TEXT,
    condition TEXT,
    purchase_price REAL,
    list_price REAL,
    date_acquired TEXT, 
    current_location TEXT,
    quantity INTEGER
)
"""

cursor.execute(create_table_query)
conn.commit()
print("Table 'books' created successfully (or already exists).")
```

### Import Book Data

```python
# Read the CSV into a pandas DataFrame
df = pd.read_csv("books.csv")

# Write the DataFrame to the database table
df.to_sql("books", conn, if_exists='replace', index=False)  # if_exists='append' if adding

conn.commit()
print("CSV data imported to 'books' table successfully.")
```
